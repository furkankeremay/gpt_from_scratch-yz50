```python
#TASK3
inputs = [0.5, 0.2]

weights = [
    [0.2, -0.1],
    [-0.5, 0.3],
    [0.8, -0.2],
    [0.6, -0.6],
]

biases = [0.1, 0.5, -0.2, 0.9]

targets = [0.7, 0.4, 0.3, 0.68]

conc = [] 

difference = []

for i in range(len(weights)): 
    b = 0
    for n in range(len(inputs)): 
        a = inputs[n] * weights[i][n] 
        b = b + a
        
    last = b + biases[i]
    conc.append(last)


for i in range(len(targets)):
    square = (targets[i]-conc[i])**2
    difference.append(square)
final = (sum(difference))/len(difference)
print(final)
```

```python
#TASK4
import matplotlib.pyplot as plt
import numpy as np

inputs = 0.5
weights = np.arange(-10,10.25,0.25)
bias = 3
target = 4
square=[]

for i in range(len(weights)):
    result = inputs*weights[i]+bias
    diff = (result-target)**2
    square.append(diff)
plt.plot(weights,square)
```

```python
#TASK5
inputs = 0.5
target = 5
bias = 2
h = 0.0001
rate = 0.1
weight = 0
for i in range(301):
    pred1 = (inputs*weight)+bias
    loss1 = (pred1-target)**2
    pred2 = (inputs*(weight+h))+bias
    loss2 = (pred2-target)**2
    grad = (loss2-loss1)/h
    weight = weight - (rate*grad)
    if i % 30 == 0:
        print(f"Adım: {i}, Ağırlık: {weight:.4f}, Hata: {loss1:.4f}")
```
